# Python Package Ecosystem

Starter notebook for the Kaggle dataset in this folder. It loads `pypi_packages.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `pypi_packages.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("pypi_packages.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Download concentration

```python
top = df.sort_values("monthly_downloads", ascending=False).head(20)
top.set_index("project")["monthly_downloads"].plot(kind="bar", figsize=(10, 4), title="Top 20 by 30-day downloads")
```

## 5. Licence mix

```python
df.groupby("license_class")["monthly_downloads"].agg(["count", "sum"]).sort_values("sum", ascending=False)
```

## 6. Release cadence

```python
df["release_cadence"].value_counts().to_frame("packages")
```

## 7. Dependencies vs downloads

```python
sub = df.dropna(subset=["dependency_count", "monthly_downloads"])
sub.plot.scatter(x="dependency_count", y="monthly_downloads", s=4, alpha=0.3, logy=True, figsize=(8, 5))
```

## 8. Staleness

```python
df.nlargest(15, "days_since_release")[["project", "days_since_release", "monthly_downloads", "development_status"]]
```

## 9. Packaging completeness

```python
df.groupby("has_wheel")["monthly_downloads"].median().to_frame("median_downloads")
```
