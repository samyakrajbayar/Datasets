# U.S. Severe Weather Events 2000–2026

Starter notebook for the Kaggle dataset in this folder. It loads `severe_weather_events.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `severe_weather_events.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("severe_weather_events.csv", low_memory=False, parse_dates=["begin_datetime", "end_datetime"])
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Events by type

```python
df["event_type_group"].value_counts().head(12).to_frame("events")
```

## 5. Fatalities by event family

```python
(df.groupby("event_type_group")["total_deaths"].sum()
   .sort_values(ascending=False).head(12).to_frame("deaths"))
```

## 6. Damage over time

```python
yearly = df.groupby("year")["damage_total_usd"].sum() / 1e9
yearly.plot(kind="bar", figsize=(12, 4), title="Reported damage by year (US$ bn)")
```

## 7. Tornado path lengths

```python
tor = df[(df["is_tornado"] == "true") & df["path_length_km"].notna()]
tor["path_length_km"].clip(upper=200).hist(bins=40, figsize=(8, 4))
```

## 8. Seasonality

```python
df.pivot_table(index="month", columns="event_type_group", values="event_id",
               aggfunc="count").fillna(0).astype(int)
```
