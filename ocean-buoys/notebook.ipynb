# NOAA Ocean Buoy Network — Stations and Live Observations

Starter notebook for the Kaggle dataset in this folder. It loads `ndbc_buoys.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `ndbc_buoys.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("ndbc_buoys.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Platforms by type

```python
df["type"].value_counts().to_frame("stations")
```

## 5. Live vs silent stations

```python
df[["has_obs", "is_recent", "has_wave_data", "has_wind"]].apply(lambda s: (s.astype(str).str.lower() == "true").sum())
```

## 6. Wave height distribution

```python
df["wave_height_m"].dropna().clip(upper=8).hist(bins=40, figsize=(8, 4))
```

## 7. Water temperature vs latitude

```python
df.dropna(subset=["water_temp_c", "lat"]).plot.scatter("lat", "water_temp_c", alpha=0.4, figsize=(8, 5))
```
