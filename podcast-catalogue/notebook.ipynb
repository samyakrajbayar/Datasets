# Podcast Catalogue

Starter notebook for the Kaggle dataset in this folder. It loads `podcasts.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `podcasts.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("podcasts.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Podcasts by primary genre

```python
df["primary_genre"].value_counts().head(15).to_frame("podcasts")
```

## 5. Episode counts

```python
df["track_count"].dropna().clip(upper=1000).hist(bins=50, figsize=(8, 4))
```

## 6. Free vs paid

```python
df["is_free"].value_counts(normalize=True).mul(100).round(1)
```

## 7. Episodes per year

```python
df["episodes_per_year"].dropna().clip(upper=500).hist(bins=40, figsize=(8, 4))
```

## 8. Release recency

```python
df["release_year"].value_counts().sort_index().tail(12).plot(kind="bar", figsize=(9, 4))
```
