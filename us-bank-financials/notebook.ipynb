# US Bank Financial Statements — FDIC Institutions

Starter notebook for the Kaggle dataset in this folder. It loads `us_banks.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `us_banks.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("us_banks.csv", low_memory=False, dtype={"fdic_cert": str, "zip": str})
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Largest banks

```python
df.nlargest(15, "assets_billions")[["name", "state", "assets_billions", "deposits_billions", "employees"]]
```

## 5. Asset distribution

```python
import numpy as np
df["assets_billions"].dropna().clip(lower=0.001).hist(bins=60, log=True, figsize=(10, 4))
```

## 6. Return on assets by size band

```python
band = pd.cut(df["assets_billions"], [0, 0.1, 1, 10, 100, 10000], labels=["<100M", "100M-1B", "1-10B", "10-100B", "100B+"])
df.groupby(band, observed=True)["return_on_assets_pct"].median()
```

## 7. Uninsured deposits

```python
df.nlargest(15, "uninsured_deposit_share_pct")[["name", "assets_billions", "uninsured_deposit_share_pct"]]
```

## 8. Banks per state

```python
df["state"].value_counts().head(20).plot(kind="bar", figsize=(11, 4))
```

## 9. Loan book composition

```python
cols = ["residential_loans_usd", "commercial_industrial_loans_usd", "nonresidential_loans_usd", "consumer_loans_usd", "multifamily_loans_usd"]
(df[cols].sum() / 1e12).sort_values().plot(kind="barh", figsize=(8, 4))
```
