# Global Flight Routes Network

Starter notebook for the Kaggle dataset in this folder. It loads `flight_routes.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `flight_routes.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("flight_routes.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Busiest origin airports

```python
df["source_airport"].value_counts().head(15).to_frame("routes")
```

## 5. Airline route counts

```python
df["airline_name"].value_counts().head(15).to_frame("routes")
```

## 6. Route distance distribution

```python
df["distance_km"].dropna().clip(upper=15000).hist(bins=50, figsize=(8, 4))
```

## 7. Domestic vs international

```python
df["is_international"].value_counts().to_frame("routes")
```
