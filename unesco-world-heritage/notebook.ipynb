# UNESCO World Heritage List

Starter notebook for the Kaggle dataset in this folder. It loads `world_heritage_sites.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `world_heritage_sites.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("world_heritage_sites.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Sites by category

```python
df["category"].value_counts().to_frame("sites")
```

## 5. Sites by inscription decade

```python
df["inscription_decade"].value_counts().sort_index().to_frame("sites")
```

## 6. Sites per region

```python
df["regions"].value_counts().to_frame("sites")
```

## 7. Danger list

```python
df[df["is_in_danger"].astype(str).str.lower() == "true"][["site", "states", "inscription_year"]].head(20)
```
