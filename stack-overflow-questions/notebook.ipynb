# Stack Overflow Questions

Starter notebook for the Kaggle dataset in this folder. It loads `stackoverflow_questions.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `stackoverflow_questions.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("stackoverflow_questions.csv", low_memory=False, parse_dates=["creation_date"])
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Top tags

```python
df["tags"].dropna().str.split(";").explode().str.strip().value_counts().head(20).to_frame("questions")
```

## 5. Answered share

```python
df["is_answered"].value_counts(normalize=True).mul(100).round(1)
```

## 6. Score distribution

```python
df["score"].clip(-5, 50).hist(bins=50, figsize=(8, 4))
```

## 7. Views vs answers

```python
sub = df.dropna(subset=["view_count", "answer_count"])
sub.plot.scatter(x="view_count", y="answer_count", s=4, alpha=0.2, figsize=(8, 5))
```

## 8. Body length vs score

```python
df.plot.scatter(x="body_text_length", y="score", s=3, alpha=0.2, figsize=(8, 5))
```
