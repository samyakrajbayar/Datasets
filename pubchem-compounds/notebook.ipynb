# PubChem Compound Property Table

Starter notebook for the Kaggle dataset in this folder. It loads `pubchem_compounds.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `pubchem_compounds.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("pubchem_compounds.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Molecular weight distribution

```python
df["molecular_weight"].dropna().clip(upper=1000).hist(bins=50, figsize=(8, 4))
```

## 5. Lipinski violations

```python
df["rule_of_five_violations"].value_counts().sort_index().to_frame("compounds")
```

## 6. Mass vs lipophilicity

```python
df.dropna(subset=["molecular_weight", "xlogp"]).sample(5000, random_state=0).plot.scatter("molecular_weight", "xlogp", alpha=0.3, figsize=(8, 5))
```

## 7. Element composition

```python
df[["carbon_atoms", "nitrogen_atoms", "oxygen_atoms", "sulfur_atoms", "halogen_atoms"]].describe().T
```
