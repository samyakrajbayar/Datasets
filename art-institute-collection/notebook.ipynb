# Art Institute of Chicago Collection

Starter notebook for the Kaggle dataset in this folder. It loads `artworks.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `artworks.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("artworks.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Objects by department

```python
df["department_title"].value_counts().head(12).to_frame("objects")
```

## 5. Creation dates

```python
df["creation_year"].dropna().loc[lambda s: s.between(1000, 2030)].plot(
    kind="hist", bins=60, figsize=(10, 4), title="Objects by creation year")
```

## 6. Medium mix

```python
df["medium_class"].value_counts().head(12).to_frame("objects")
```

## 7. Colourfulness by century

```python
df.groupby("century")["colorfulness"].median().dropna().sort_values().tail(12)
```

## 8. Public domain share

```python
df.groupby("department_title")["is_public_domain"].mean().mul(100).round(1).sort_values(ascending=False)
```

## 9. Most prolific artists

```python
df["artist_title"].dropna().value_counts().head(15).to_frame("works")
```

## 10. Styles over time

```python
pivot = (df.dropna(subset=["creation_decade", "style_title"])
         .query("1850 <= creation_decade <= 1990")
         .groupby(["creation_decade", "style_title"]).size().unstack(fill_value=0))
pivot.iloc[:, :12].plot(figsize=(11, 5))
```
