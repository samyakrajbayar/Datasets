# NOAA Tide & Current Station Network

Starter notebook for the Kaggle dataset in this folder. It loads `noaa_stations.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `noaa_stations.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("noaa_stations.csv", dtype={"station_id": str}, low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Stations by state

```python
df["state"].value_counts().head(15).to_frame("stations")
```

## 5. Capability mix

```python
for c in ["is_tide_station", "is_current_station", "is_met_station", "is_water_level_station"]:
    print(c, int((df[c].astype(str).str.lower() == "true").sum()))
```

## 6. Products per station

```python
df["product_count"].value_counts().sort_index().to_frame("stations")
```

## 7. Map

```python
sub = df.dropna(subset=["latitude", "longitude"])
sub.plot.scatter(x="longitude", y="latitude", s=8, alpha=0.6, figsize=(10, 6))
```

## 8. Region summary

```python
df.groupby("state_region")["product_count"].agg(["count", "mean"]).round(2)
```
