# Magic: The Gathering Card Catalogue

Starter notebook for the Kaggle dataset in this folder. It loads `mtg_cards.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `mtg_cards.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("mtg_cards.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Rarity mix

```python
df["rarity"].value_counts().to_frame("cards")
```

## 5. Card types

```python
for c in ["is_creature", "is_land", "is_instant", "is_sorcery", "is_artifact", "is_enchantment", "is_planeswalker"]:
    print(c, int((df[c].astype(str).str.lower() == "true").sum()))
```

## 6. Mana cost distribution

```python
df["cmc"].dropna().clip(upper=12).value_counts().sort_index().plot(kind="bar", figsize=(9, 4))
```

## 7. Price by rarity

```python
df.boxplot(column="price_usd", by="rarity", figsize=(9, 4))
```

## 8. Colours

```python
df["color_identity"].fillna("C").value_counts().head(12).to_frame("cards")
```
