# U.S. Nursing Homes — Ownership, Capacity and CMS Ratings

Starter notebook for the Kaggle dataset in this folder. It loads `nursing_homes.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `nursing_homes.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("nursing_homes.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Ownership mix

```python
df["ownership_group"].value_counts().to_frame("providers")
```

## 5. Overall rating distribution

```python
df["overall_rating"].value_counts().sort_index().to_frame("providers")
```

## 6. Ratings by ownership

```python
df.groupby("ownership_group")["overall_rating"].agg(["count", "mean"]).round(2)
```

## 7. Occupancy vs rating

```python
df.dropna(subset=["occupancy_rate", "overall_rating"]).plot.scatter("overall_rating", "occupancy_rate", alpha=0.2, figsize=(8, 4))
```
