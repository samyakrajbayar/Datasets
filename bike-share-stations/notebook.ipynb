# Bike-Share Station Network — North America

Starter notebook for the Kaggle dataset in this folder. It loads `bike_share_stations.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `bike_share_stations.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("bike_share_stations.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Stations per system

```python
df.groupby("system").agg(stations=("station_id", "count"), docks=("capacity", "sum")).sort_values("stations", ascending=False)
```

## 5. Capacity distribution

```python
df["capacity"].dropna().clip(upper=80).hist(bins=40, figsize=(8, 4))
```

## 6. Payment methods

```python
for c in ["accepts_key", "accepts_creditcard", "accepts_phone", "has_kiosk"]:
    print(c, round((df[c].astype(str).str.lower() == "true").mean() * 100, 1), "%")
```

## 7. Largest stations

```python
df.nlargest(10, "capacity")[["system", "name", "capacity"]]
```
