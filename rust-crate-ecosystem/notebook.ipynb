# Rust Crate Ecosystem

Starter notebook for the Kaggle dataset in this folder. It loads `rust_crates.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `rust_crates.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("rust_crates.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Download distribution

```python
df["downloads"].dropna().clip(upper=10_000_000).hist(bins=50, figsize=(9, 4), log=True)
```

## 5. Release cadence

```python
df["versions_per_year"].dropna().clip(upper=50).hist(bins=40, figsize=(8, 4))
```

## 6. Maintenance

```python
df["days_since_update"].dropna().clip(upper=2000).hist(bins=50, figsize=(8, 4))
```

## 7. Domain flags

```python
for c in ["is_async", "is_cli", "is_wasm", "is_embedded", "is_proc_macro", "is_serialization"]:
    print(c, int((df[c].astype(str).str.lower() == "true").sum()))
```

## 8. Downloads per day vs age

```python
sub = df.dropna(subset=["age_days", "downloads_per_day"])
sub.plot.scatter(x="age_days", y="downloads_per_day", s=3, alpha=0.2, logy=True, figsize=(8, 5))
```
