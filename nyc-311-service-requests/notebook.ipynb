# NYC 311 Service Requests

Starter notebook for the Kaggle dataset in this folder. It loads `nyc_311_requests.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `nyc_311_requests.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("nyc_311_requests.csv", low_memory=False, parse_dates=["created_date", "closed_date"])
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Top complaint types

```python
df["complaint_type"].value_counts().head(15).to_frame("requests")
```

## 5. Requests by borough

```python
df["borough"].value_counts().to_frame("requests")
```

## 6. Response time distribution

```python
df["response_days"].dropna().clip(upper=30).hist(bins=40, figsize=(9, 4))
```

## 7. Complaint categories

```python
df["complaint_category"].value_counts().to_frame("requests")
```

## 8. Hourly pattern

```python
df.groupby("hour")["unique_key"].count().plot(kind="bar", figsize=(10, 4))
```
