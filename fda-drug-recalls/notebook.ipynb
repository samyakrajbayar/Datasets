# FDA Drug Recalls — Enforcement Reports

Starter notebook for the Kaggle dataset in this folder. It loads `fda_drug_recalls.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `fda_drug_recalls.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("fda_drug_recalls.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Classification mix

```python
df["classification"].value_counts().to_frame("recalls")
```

## 5. Recalls by year

```python
df["year"].value_counts().sort_index().plot(kind="bar", figsize=(11, 4))
```

## 6. Reason categories

```python
df["reason_category"].value_counts().to_frame("recalls")
```

## 7. Processing lag

```python
df["days_initiation_to_classification"].dropna().clip(0, 1000).hist(bins=40, figsize=(8, 4))
```

## 8. Top firms

```python
df["recalling_firm"].value_counts().head(15).to_frame("recalls")
```
