# FDA Food & Supplement Recalls

Starter notebook for the Kaggle dataset in this folder. It loads `food_recalls.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `food_recalls.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("food_recalls.csv", low_memory=False, parse_dates=["recall_initiation_date"])
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Classification mix

```python
df["classification"].value_counts().to_frame("recalls")
```

## 5. Recalls per year

```python
df["initiation_year"].value_counts().sort_index().to_frame("recalls")
```

## 6. Reason keywords

```python
for c in ["mentions_undeclared_allergen", "mentions_listeria", "mentions_salmonella", "mentions_foreign_material", "mentions_labeling"]:
    print(c, int((df[c].astype(str).str.lower() == "true").sum()))
```

## 7. Recall duration

```python
df["recall_duration_days"].dropna().clip(upper=1000).hist(bins=40, figsize=(8, 4))
```
