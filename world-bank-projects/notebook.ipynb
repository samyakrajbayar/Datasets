# World Bank Project Portfolio

Starter notebook for the Kaggle dataset in this folder. It loads `world_bank_projects.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `world_bank_projects.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("world_bank_projects.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Commitments by region

```python
pd.to_numeric(df["totalamt"], errors="coerce").groupby(df["regionname"]).sum().sort_values(ascending=False).div(1e9).round(2).to_frame("bn_usd")
```

## 5. Projects by status

```python
df["projectstatusdisplay"].value_counts().to_frame("projects")
```

## 6. Commitments by approval year

```python
pd.to_numeric(df["totalamt"], errors="coerce").groupby(df["approval_year"]).sum().div(1e9).round(2).tail(30).plot(figsize=(10, 4))
```

## 7. Top sectors

```python
df["sector1"].value_counts().head(15).to_frame("projects")
```
