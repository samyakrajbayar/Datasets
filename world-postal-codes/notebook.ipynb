# World Postal Code Directory

Starter notebook for the Kaggle dataset in this folder. It loads `postal_codes.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `postal_codes.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("postal_codes.csv", dtype={"postal_code": str}, low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Records by country

```python
df["country_name"].value_counts().head(15).to_frame("postal_records")
```

## 5. Postal code length distribution

```python
df["postal_length"].value_counts().sort_index().head(12).to_frame("codes")
```

## 6. Numeric vs alphanumeric

```python
df["is_numeric_postal"].value_counts(normalize=True).mul(100).round(1)
```

## 7. Administrative depth

```python
df["admin_depth"].value_counts().sort_index().to_frame("records")
```

## 8. Map sample

```python
sample = df.dropna(subset=["latitude", "longitude"]).sample(5000, random_state=0)
sample.plot.scatter(x="longitude", y="latitude", s=1, alpha=0.3, figsize=(11, 6))
```
