# Global Health Panel

Starter notebook for the Kaggle dataset in this folder. It loads `who_health_panel.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `who_health_panel.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("who_health_panel.csv", low_memory=False)
df["life_expectancy"] = pd.to_numeric(df["life_expectancy"], errors="coerce")
print(df.shape)
```

## 2. Shape, types and missing values

```python
import pandas as pd

print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Coverage of the panel

```python
df.groupby("year")["n_indicators_reported"].mean().round(1).plot(figsize=(9, 4))
```

## 5. Life expectancy by region over time

```python
(df.dropna(subset=["life_expectancy"])
   .groupby(["year", "region"])["life_expectancy"].mean().unstack()
   .plot(figsize=(11, 5)))
```

## 6. Life expectancy vs health spending

```python
sub = df.dropna(subset=["health_spend_per_capita_usd", "life_expectancy"]).query("year >= 2015")
sub.plot.scatter(x="health_spend_per_capita_usd", y="life_expectancy", logx=True, s=8, figsize=(9, 5))
```

## 7. Workforce density and child mortality

```python
sub = df.dropna(subset=["doctors_per_1000", "under5_mortality_per_1000"]).query("year >= 2015")
sub.plot.scatter(x="doctors_per_1000", y="under5_mortality_per_1000", logy=True, s=8, figsize=(9, 5))
```

## 8. Country profile snapshot

```python
latest = df[df["year"] == df["year"].max()]
latest.nlargest(10, "life_expectancy")[["country_name", "life_expectancy", "healthy_life_expectancy",
                                       "health_spend_pct_gdp", "doctors_per_1000"]]
```

## 9. Missingness by indicator

```python
cols = [c for c in df.columns if df[c].dtype != object]
df[cols].isna().mean().mul(100).sort_values(ascending=False).round(1).head(20)
```
