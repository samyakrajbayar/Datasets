# Open-Access Journals — DOAJ Directory

Starter notebook for the Kaggle dataset in this folder. It loads `doaj_journals.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `doaj_journals.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("doaj_journals.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Journals by country

```python
df["country"].value_counts().head(15).to_frame("journals")
```

## 5. APC presence

```python
df["has_apc"].value_counts(normalize=True).mul(100).round(1)
```

## 6. Licence mix

```python
df["license_attributes"].dropna().str.split(";").explode().str.strip().value_counts().head(10)
```

## 7. Publication speed

```python
df["weeks_to_publication_numeric"].dropna().clip(upper=52).hist(bins=30, figsize=(8, 4))
```

## 8. Policy coverage

```python
for c in ["has_preservation", "has_deposit_policy", "has_persistent_identifier", "has_plagiarism_policy"]:
    print(c, round((df[c].astype(str).str.lower() == "true").mean() * 100, 1), "%")
```
