# Drug-Like Molecules — ChEMBL Physicochemistry

Starter notebook for the Kaggle dataset in this folder. It loads `chembl_molecules.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `chembl_molecules.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("chembl_molecules.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Clinical phase

```python
df["max_phase"].value_counts().sort_index().to_frame("molecules")
```

## 5. Molecular weight vs ALogP

```python
sub = df.dropna(subset=["mw_freebase", "alogp"])
sub.plot.scatter(x="mw_freebase", y="alogp", s=4, alpha=0.2, figsize=(8, 5))
```

## 6. Lipinski violations

```python
df["num_ro5_violations"].value_counts().sort_index().to_frame("molecules")
```

## 7. Approved vs preclinical properties

```python
df.groupby("is_approved")[["mw_freebase", "alogp", "psa", "qed_weighted"]].median().round(2)
```

## 8. Modalities

```python
df["molecule_type"].value_counts().head(10).to_frame("molecules")
```
