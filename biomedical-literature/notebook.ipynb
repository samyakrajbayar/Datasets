# Biomedical Literature 2024–2025

Starter notebook for the Kaggle dataset in this folder. It loads `europepmc_articles.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `europepmc_articles.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("europepmc_articles.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Publication years

```python
df["pub_year"].value_counts().sort_index().to_frame("articles")
```

## 5. Top journals

```python
df["journal_title"].value_counts().head(15).to_frame("articles")
```

## 6. Review share

```python
df["is_review"].value_counts(normalize=True).mul(100).round(1)
```

## 7. Citations distribution

```python
df["cited_by_count"].dropna().clip(upper=100).hist(bins=40, figsize=(8, 4))
```

## 8. MeSH terms

```python
(df["mesh_terms"].dropna().str.split(";").explode()
   .str.strip().value_counts().head(20).to_frame("articles"))
```
