# Holocene Volcanoes

Starter notebook for the Kaggle dataset in this folder. It loads `volcanoes.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `volcanoes.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("volcanoes.csv", low_memory=False)
df["last_eruption_year"] = pd.to_numeric(df["last_eruption_year"], errors="coerce")
print(df.shape)
```

## 2. Shape, types and missing values

```python
import pandas as pd

print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. The Ring of Fire

Every volcano plotted by longitude and latitude.

```python
df.plot.scatter(x="longitude", y="latitude", s=6, alpha=0.6, figsize=(13, 6))
```

## 5. Volcano types

```python
df["volcano_type_group"].value_counts().head(10).plot.bar(figsize=(8, 4))
```

## 6. Eruption recency

```python
df["eruption_recency_class"].value_counts().plot.pie(figsize=(6, 6), autopct="%1.1f%%")
```

## 7. Elevation distribution by latitude band

```python
df.boxplot(column="elevation_m", by="latitude_band", figsize=(12, 5), rot=30)
```

## 8. Most active countries since 1500

```python
(df[df["is_historical_eruption"].astype(str).str.lower() == "true"]
   .groupby("country").size().sort_values(ascending=False).head(15).plot.barh(figsize=(8, 6)))
```

## 9. Tectonic settings

```python
df["tectonic_setting"].value_counts().head(10)
```
