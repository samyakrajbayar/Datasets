# Live Air Traffic Snapshot — Global Aircraft States

Starter notebook for the Kaggle dataset in this folder. It loads `aircraft_states.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `aircraft_states.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("aircraft_states.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Aircraft per country

```python
df["origin_country"].value_counts().head(20).to_frame("aircraft")
```

## 5. Altitude distribution

```python
df["altitude_ft"].dropna().clip(upper=45000).hist(bins=45, figsize=(8, 4))
```

## 6. Airborne vs on ground

```python
df["is_ground"].value_counts().to_frame("aircraft")
```

## 7. Speed vs altitude

```python
df.dropna(subset=["altitude_ft", "speed_kmh"]).plot.scatter("altitude_ft", "speed_kmh", alpha=0.3, figsize=(8, 5))
```
