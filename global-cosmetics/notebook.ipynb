# Global Cosmetics & Personal Care

Starter notebook for the Kaggle dataset in this folder. It loads `cosmetics.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `cosmetics.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("cosmetics.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Products by category group

```python
df["category_group"].value_counts().to_frame("products")
```

## 5. Top brands

```python
df["brands"].value_counts().head(15).to_frame("products")
```

## 6. Claim prevalence

```python
for c in ["is_vegan", "is_organic", "is_natural", "is_hypoallergenic", "is_paraben_free", "is_sulfate_free"]:
    print(c, round((df[c].astype(str).str.lower() == "true").mean() * 100, 1), "%")
```

## 7. Ingredient count

```python
df["ingredient_count"].dropna().clip(upper=60).hist(bins=40, figsize=(8, 4))
```

## 8. Category vs ingredient complexity

```python
df.groupby("category_group")["ingredient_count"].median().sort_values().to_frame("median_ingredients")
```
