# UK Property Sales 2025–2026

Starter notebook for the Kaggle dataset in this folder. It loads `uk_property_sales.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `uk_property_sales.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("uk_property_sales.csv", low_memory=False, parse_dates=["date_of_transfer"])
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Median price by property type

```python
df.groupby("property_type")["price"].agg(["count", "median"]).sort_values("median", ascending=False)
```

## 5. Monthly sales volume

```python
df.groupby(df["date_of_transfer"].dt.to_period("M")).size().plot(figsize=(11, 4))
```

## 6. Price distribution

```python
df["price"].clip(upper=1500000).hist(bins=60, figsize=(8, 4))
```

## 7. New build share by quarter

```python
df.groupby("transfer_quarter")["is_new_build"].apply(lambda s: (s.astype(str).str.lower() == "true").mean())
```
