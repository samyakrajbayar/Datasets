# U.S. Hospital Quality — CMS Star Ratings

Starter notebook for the Kaggle dataset in this folder. It loads `hospitals.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `hospitals.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("hospitals.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Hospitals by type

```python
df["hospital_type"].value_counts().to_frame("hospitals")
```

## 5. Ownership mix

```python
df["ownership_group"].value_counts().to_frame("hospitals")
```

## 6. Overall rating distribution

```python
df["hospital_overall_rating"].value_counts().sort_index().to_frame("hospitals")
```

## 7. Ratings by ownership

```python
pd.to_numeric(df["hospital_overall_rating"], errors="coerce").groupby(df["ownership_group"]).agg(["count", "mean"]).round(2)
```
