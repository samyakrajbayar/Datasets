# Human Proteome — Reviewed Protein Annotations

Starter notebook for the Kaggle dataset in this folder. It loads `human_proteins.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `human_proteins.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("human_proteins.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Length distribution

```python
df["length"].dropna().clip(upper=3000).hist(bins=60, figsize=(9, 4))
```

## 5. Enzymes by EC class

```python
df["ec_class"].value_counts().sort_index().to_frame("proteins")
```

## 6. Localisation flags

```python
for c in ["is_secreted", "is_membrane_protein", "is_nuclear", "is_mitochondrial"]:
    print(c, round((df[c].astype(str).str.lower() == "true").mean() * 100, 1), "%")
```

## 7. Structure coverage

```python
df["has_structure"].value_counts(normalize=True).mul(100).round(1)
```

## 8. Annotation richness

```python
df[["function_length", "domain_count", "go_term_count", "variant_count"]].describe().T.round(2)
```
