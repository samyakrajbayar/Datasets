# Deezer Music Catalogue Sample

Starter notebook for the Kaggle dataset in this folder. It loads `deezer_tracks.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `deezer_tracks.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("deezer_tracks.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Tracks per query

```python
df["query"].value_counts().head(20).to_frame("tracks")
```

## 5. Duration distribution

```python
df["duration_minutes"].dropna().clip(upper=10).hist(bins=50, figsize=(8, 4))
```

## 6. Popularity by genre query

```python
df.groupby("query")["rank"].median().sort_values().tail(15).to_frame("median_rank")
```

## 7. Explicit share

```python
df.groupby("query")["is_explicit"].apply(lambda s: (s.astype(str).str.lower() == "true").mean()).sort_values().tail(12)
```

## 8. Top artists

```python
df["artist_name"].value_counts().head(20).to_frame("tracks")
```
