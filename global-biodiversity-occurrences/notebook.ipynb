# Global Biodiversity Occurrences

Starter notebook for the Kaggle dataset in this folder. It loads `occurrences.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `occurrences.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("occurrences.csv", low_memory=False)
print(df.shape)
df[["scientificName", "kingdom", "country_name", "decimalLatitude", "year"]].head()
```

## 2. Shape, types and missing values

```python
import pandas as pd

print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Records per kingdom

```python
df["kingdom"].value_counts().head(10).plot.bar(figsize=(8, 4))
```

## 5. Latitudinal diversity gradient

Records per 5-degree band of absolute latitude.

```python
bins = range(0, 95, 5)
band = pd.cut(df["abs_latitude"], bins=bins)
band.value_counts().sort_index().plot.bar(figsize=(11, 4))
```

## 6. Sampling bias between countries

```python
df.groupby("country_name").agg(records=("key", "count"),
                                    species=("species", "nunique"),
                                    families=("family", "nunique")).sort_values("records", ascending=False)
```

## 7. Data quality flags

```python
df["n_issues"].value_counts().sort_index().plot.bar(figsize=(8, 4))
df["has_issues"].value_counts(normalize=True).round(3)
```

## 8. Record type mix by region

```python
(df.groupby(["region", "basis_group"]).size()
   .unstack(fill_value=0).pipe(lambda t: t.div(t.sum(axis=1), axis=0) * 100).round(1))
```

## 9. Observation years

```python
df["year"].dropna().astype(int).plot.hist(bins=40, figsize=(10, 4))
```
