# Space Weather Time Series — X-Ray Flux, Kp and F10.7

Starter notebook for the Kaggle dataset in this folder. It loads `space_weather.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `space_weather.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("space_weather.csv", low_memory=False, parse_dates=["timestamp"])
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. X-ray flux over time

```python
df.set_index("timestamp")["xray_flux"].plot(figsize=(11, 4), logy=True)
```

## 5. Flare classes

```python
df["xray_class"].value_counts().to_frame("minutes")
```

## 6. Kp distribution

```python
df["kp_index"].dropna().value_counts().sort_index().to_frame("minutes")
```

## 7. Flux vs Kp

```python
df.dropna(subset=["kp_index", "xray_flux"]).plot.scatter("kp_index", "xray_flux", alpha=0.2, figsize=(8, 4))
```
