# Chicago Crime Incidents

Starter notebook for the Kaggle dataset in this folder. It loads `chicago_crimes.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `chicago_crimes.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("chicago_crimes.csv", low_memory=False, parse_dates=["date"])
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Crime categories

```python
df["crime_category"].value_counts().to_frame("incidents")
```

## 5. Top primary types

```python
df["primary_type"].value_counts().head(15).to_frame("incidents")
```

## 6. Arrest rate by category

```python
(df.assign(arrest=df["is_arrest"].astype(str).str.lower() == "true")
   .groupby("crime_category")["arrest"].mean().mul(100).round(1).to_frame("arrest_%"))
```

## 7. Hourly pattern

```python
df.groupby("hour")["id"].count().plot(kind="bar", figsize=(10, 4))
```

## 8. Map sample

```python
sub = df.dropna(subset=["latitude", "longitude"]).sample(8000, random_state=0)
sub.plot.scatter(x="longitude", y="latitude", s=1, alpha=0.25, figsize=(8, 9))
```
