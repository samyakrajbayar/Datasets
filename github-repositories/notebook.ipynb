# Most-Starred GitHub Repositories

Starter notebook for the Kaggle dataset in this folder. It loads `github_repositories.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `github_repositories.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("github_repositories.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Stars by language

```python
df.groupby("language").agg(repos=("name", "count"), stars=("stars", "sum")).sort_values("stars", ascending=False).head(15)
```

## 5. Star distribution

```python
df["stars"].clip(upper=300000).hist(bins=50, figsize=(8, 4))
```

## 6. Licence mix

```python
df["license"].value_counts().head(12).to_frame("repos")
```

## 7. Maintenance status

```python
for c in ["is_active", "is_stale", "is_archived"]:
    print(c, round((df[c].astype(str).str.lower() == "true").mean() * 100, 1), "%")
```
