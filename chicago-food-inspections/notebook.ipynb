# Chicago Food Inspections 2023–2026

Starter notebook for the Kaggle dataset in this folder. It loads `chicago_food_inspections.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `chicago_food_inspections.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("chicago_food_inspections.csv", low_memory=False, parse_dates=["inspection_date"])
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Inspection results

```python
df["results"].value_counts().head(10).to_frame("inspections")
```

## 5. Facility groups

```python
df["facility_group"].value_counts().to_frame("inspections")
```

## 6. Failure rate by risk

```python
df.groupby("risk")["is_fail"].apply(lambda s: (s.astype(str).str.lower() == "true").mean()).round(3)
```

## 7. Violations per inspection

```python
df["violation_count"].clip(upper=30).hist(bins=30, figsize=(8, 4))
```
