# Airbnb Listings — Eight World Cities

Starter notebook for the Kaggle dataset in this folder. It loads `airbnb_listings.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `airbnb_listings.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("airbnb_listings.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Listings per city

```python
df["city"].value_counts().to_frame("listings")
```

## 5. Room type mix

```python
pd.crosstab(df["city"], df["room_type"])
```

## 6. Nightly price distribution

```python
df["price_numeric"].dropna().clip(upper=600).hist(bins=60, figsize=(8, 4))
```

## 7. Median price by city and room type

```python
df.pivot_table(index="city", columns="room_type", values="price_numeric", aggfunc="median").round(0)
```
