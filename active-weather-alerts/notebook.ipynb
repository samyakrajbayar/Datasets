# Active U.S. Weather Alerts — Live NWS Snapshot

Starter notebook for the Kaggle dataset in this folder. It loads `weather_alerts.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `weather_alerts.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("weather_alerts.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Alerts by event

```python
df["event"].value_counts().head(20).to_frame("alerts")
```

## 5. Severity mix

```python
df["severity"].value_counts().to_frame("alerts")
```

## 6. Hazard flags

```python
for c in ["is_warning", "is_watch", "is_flood", "is_winter", "is_fire", "is_severe_storm"]:
    print(c, int((df[c].astype(str).str.lower() == "true").sum()))
```

## 7. Alert duration

```python
df["duration_hours"].dropna().clip(upper=72).hist(bins=40, figsize=(8, 4))
```
