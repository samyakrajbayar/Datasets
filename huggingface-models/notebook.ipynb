# Hugging Face Model Hub Catalogue

Starter notebook for the Kaggle dataset in this folder. It loads `hf_models.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `hf_models.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("hf_models.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Models by pipeline

```python
df["pipeline_tag"].value_counts().head(20).to_frame("models")
```

## 5. Downloads by task

```python
df.groupby("pipeline_tag")["downloads"].sum().sort_values(ascending=False).head(15).to_frame("downloads")
```

## 6. Licence mix

```python
df["license"].value_counts().head(15).to_frame("models")
```

## 7. Downloads vs likes

```python
df.dropna(subset=["downloads", "likes"]).plot.scatter("downloads", "likes", alpha=0.3, loglog=True, figsize=(8, 5))
```
