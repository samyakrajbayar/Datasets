# U.S. Tornado Catalog 1950–2024

Starter notebook for the Kaggle dataset in this folder. It loads `tornadoes.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `tornadoes.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("tornadoes.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Tornadoes per year

```python
df.groupby("yr").size().plot(figsize=(11, 4), title="Recorded tornadoes per year")
```

## 5. Ratings

```python
df["magnitude_label"].value_counts().head(14).to_frame("tornadoes")
```

## 6. Where they happen\n\nTop 15 states by count.

```python
df["state_name"].value_counts().head(15).to_frame("tornadoes")
```

## 7. Seasonality

```python
df["month_name"].value_counts().reindex([
    "January","February","March","April","May","June","July","August","September","October","November","December"]
).plot(kind="bar", figsize=(10, 4), title="Tornadoes by month")
```

## 8. Casualties vs magnitude

```python
df.groupby("magnitude_label")[["fat", "inj"]].sum().sort_values("fat", ascending=False).head(12)
```

## 9. Path length by rating

```python
df.boxplot(column="len", by="magnitude_label", figsize=(11, 5), rot=45)
```

## 10. F vs EF era

```python
df.groupby("scale")["mag"].agg(["count", "mean", "median"]).round(2)
```
