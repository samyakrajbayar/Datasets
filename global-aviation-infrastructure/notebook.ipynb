# Global Aviation Infrastructure

Starter notebook for the Kaggle dataset in this folder. It loads `airports.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `airports.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("airports.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Airport types

```python
df["type"].value_counts().to_frame("airports")
```

## 5. Runway length distribution

```python
df["longest_runway_m"].dropna().clip(upper=5000).hist(bins=50, figsize=(9, 4))
```

## 6. Top countries by airport count

```python
df["country_name"].value_counts().head(15).to_frame("airports")
```

## 7. Paved share by region

```python
(df.assign(paved=df["paved_runway_count"] > 0)
   .groupby("country_region")["paved"].mean().mul(100).round(1).to_frame("paved_%"))
```

## 8. Scheduled service by airport size

```python
pd.crosstab(df["type"], df["has_scheduled_service"], normalize="index").round(3)
```
