# Retracted Research Papers — Retraction Watch

Starter notebook for the Kaggle dataset in this folder. It loads `retractions.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `retractions.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("retractions.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Notice types

```python
df["retraction_nature"].value_counts().head(10).to_frame("records")
```

## 5. Retractions per year

```python
years = pd.to_numeric(df["retraction_year"], errors="coerce").dropna().astype(int)
print("year range:", years.min(), years.max())
years.value_counts().sort_index().tail(40).plot(kind="bar", figsize=(12, 4))
```

## 6. Reason keywords

```python
for c in ["mentions_plagiarism", "mentions_duplication", "mentions_data_issues", "mentions_fabrication", "mentions_error", "mentions_peer_review"]:
    print(c, int((df[c].astype(str).str.lower() == "true").sum()))
```

## 7. Publication-to-retraction lag

```python
df["lag_years"].dropna().clip(upper=25).hist(bins=50, figsize=(8, 4))
```
