# Global Packaged Food

Starter notebook for the Kaggle dataset in this folder. It loads `food_products.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `food_products.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("food_products.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Nutri-Score distribution

```python
df["nutriscore_grade"].value_counts().reindex(["a", "b", "c", "d", "e"])
```

## 5. Sugar by category

```python
df.groupby("categories_tags")["sugars_100g"].median().dropna().sort_values().tail(12)
```

## 6. Ultra-processed share

```python
df["nova_group"].value_counts().sort_index().to_frame("products")
```

## 7. Nutrients vs Nutri-Score

```python
df.boxplot(column="sugars_100g", by="nutriscore_grade", figsize=(9, 4))
```

## 8. Salt vs sugar

```python
sub = df.dropna(subset=["salt_100g", "sugars_100g"]).query("salt_100g < 10 and sugars_100g < 60")
sub.plot.scatter(x="sugars_100g", y="salt_100g", s=4, alpha=0.25, figsize=(8, 5))
```

## 9. Label prevalence

```python
for col in ["is_organic", "is_vegan", "is_vegetarian", "is_gluten_free"]:
    print(col, round((df[col].astype(str).str.lower() == "true").mean() * 100, 1), "%")
```
