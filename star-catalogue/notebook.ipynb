# Star Catalogue — HYG Hipparcos/Bright Star/Gliese Merge

Starter notebook for the Kaggle dataset in this folder. It loads `hyg_stars.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `hyg_stars.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("hyg_stars.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Spectral classes

```python
df["spectral_class"].value_counts().head(10).to_frame("stars")
```

## 5. Colour-magnitude diagram

```python
df.dropna(subset=["ci", "absmag"]).sample(20000, random_state=0).plot.scatter("ci", "absmag", alpha=0.2, figsize=(8, 6))
```

## 6. Nearest stars

```python
df.nsmallest(15, "dist")[["proper", "spect", "dist", "distance_ly", "mag"]]
```

## 7. Magnitude distribution

```python
df["mag"].clip(upper=12).hist(bins=60, figsize=(8, 4))
```
