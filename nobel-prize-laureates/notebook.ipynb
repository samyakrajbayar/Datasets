# Nobel Prize Laureates

Starter notebook for the Kaggle dataset in this folder. It loads `nobel_laureates.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `nobel_laureates.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("nobel_laureates.csv")
df["age_at_award"] = pd.to_numeric(df["age_at_award"], errors="coerce")
print(df.shape)
df[["prize_year", "category", "known_name", "age_at_award"]].head()
```

## 2. Shape, types and missing values

```python
import pandas as pd

print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Prizes per decade

```python
(df.drop_duplicates(["prize_year", "category"])
   .groupby([df["decade"], "category"]).size().unstack(fill_value=0)
   .plot.bar(stacked=True, figsize=(12, 6)))
```

## 5. Age at award by category

```python
df.boxplot(column="age_at_award", by="category", figsize=(11, 5), rot=30)
```

## 6. Gender over time

```python
(df[df["laureate_type"] == "person"]
   .assign(gender=lambda t: t["gender"].fillna("unknown"))
   .groupby(["decade", "gender"]).size().unstack(fill_value=0)
   .plot.bar(stacked=True, figsize=(12, 5)))
```

## 7. Most frequent affiliations

```python
df["first_affiliation"].value_counts().head(20).sort_values().plot.barh(figsize=(8, 6))
```

## 8. Real prize value per laureate

```python
val = (df.drop_duplicates(["prize_year", "category"])
       .assign(per_laureate=lambda t: t["prize_amount_adjusted_sek"] / t["n_laureates"]))
val.sort_values("prize_year").plot(x="prize_year", y="per_laureate", figsize=(10, 5))
```
