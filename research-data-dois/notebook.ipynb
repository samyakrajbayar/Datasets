# Research Data DOIs — DataCite

Starter notebook for the Kaggle dataset in this folder. It loads `datacite_dois.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `datacite_dois.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("datacite_dois.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Resource types

```python
df["resource_type_general"].value_counts().head(12).to_frame("dois")
```

## 5. Top repositories

```python
df["publisher"].value_counts().head(15).to_frame("dois")
```

## 6. Publication years

```python
df["publication_year"].value_counts().sort_index().tail(25).plot(kind="bar", figsize=(11, 4))
```

## 7. Metadata coverage

```python
for c in ["has_orcid", "has_funding", "has_geo", "has_rights", "has_related"]:
    print(c, round((df[c].astype(str).str.lower() == "true").mean() * 100, 1), "%")
```

## 8. Views vs downloads

```python
sub = df.dropna(subset=["view_count", "download_count"])
sub.plot.scatter(x="view_count", y="download_count", s=4, alpha=0.25, figsize=(8, 5))
```
