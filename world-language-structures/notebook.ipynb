# World Language Structures — WALS Typological Atlas

Starter notebook for the Kaggle dataset in this folder. It loads `world_languages.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `world_languages.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("world_languages.csv", low_memory=False)
print(df.shape)
print("features:", sum(c.startswith("wals_") and c != "wals_id" for c in df.columns))
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Languages per macroarea

```python
df["macroarea"].value_counts().plot(kind="bar", figsize=(9, 4))
```

## 5. Largest language families

```python
df["family"].value_counts().head(15).plot(kind="bar", figsize=(11, 4))
```

## 6. A single feature

```python
df["wals_81a_order_of_subject_object_and_verb"].value_counts(dropna=False).head(12)
```

## 7. Documentation coverage

```python
df["features_answered"].hist(bins=40, figsize=(9, 4))
```

## 8. Feature matrix correlation

```python
feat = [c for c in df.columns if c.startswith("wals_")]
codes = df[feat].apply(lambda s: s.astype("category").cat.codes)
corr = codes.corr()
corr.iloc[:40, :40]
```

## 9. Word-order profile

```python
order = [c for c in df.columns if "order_of" in c][:6]
df[order].describe()
```
