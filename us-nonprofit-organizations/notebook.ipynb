# US Nonprofit Organisations — IRS Business Master File

Starter notebook for the Kaggle dataset in this folder. It loads `us_nonprofits.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `us_nonprofits.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("us_nonprofits.csv", low_memory=False, dtype={"ein": str, "zip5": str})
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. NTEE categories

```python
df["ntee_category"].value_counts().head(15).to_frame("organisations")
```

## 5. States with the most charities

```python
df["state"].value_counts().head(15).plot(kind="bar", figsize=(10, 4))
```

## 6. Asset bands

```python
order = ["0", "1-9,999", "10,000-24,999", "25,000-99,999", "100,000-499,999", "500,000-999,999", "1,000,000-4,999,999", "5,000,000-9,999,999", "10,000,000-49,999,999", "50,000,000+"]
df["asset_band"].value_counts().reindex(order).plot(kind="bar", figsize=(11, 4))
```

## 7. Ruling year

```python
pd.to_numeric(df["ruling_year"], errors="coerce").dropna().astype(int).value_counts().sort_index().plot(figsize=(11, 4))
```

## 8. Revenue by category

```python
rev = pd.to_numeric(df["revenue_amount"], errors="coerce")
df.assign(revenue=rev).groupby("ntee_category")["revenue"].median().sort_values(ascending=False).head(12)
```
