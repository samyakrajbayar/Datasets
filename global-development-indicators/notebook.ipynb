# Global Development Indicators 1960–2025

Starter notebook for the Kaggle dataset in this folder. It loads `development_indicators.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `development_indicators.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("development_indicators.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. GDP per capita vs life expectancy

```python
latest = df[df["year"] == df["year"].max()].dropna(subset=["gdp_per_capita_usd", "life_expectancy_years"])
latest.plot.scatter(x="gdp_per_capita_usd", y="life_expectancy_years", s=12, alpha=0.6,
                    logx=True, figsize=(8, 5))
```

## 5. Income groups

```python
df["income_level"].value_counts().to_frame("country_years")
```

## 6. Indicator completeness

```python
(df["indicators_present"].describe().round(1).to_frame("value"))
```

## 7. Regional averages over time

```python
(df[df["year"] >= 1990].groupby(["year", "region"])["gdp_per_capita_usd"]
   .mean().unstack().plot(logy=True, figsize=(11, 5)))
```

## 8. Urban share vs internet adoption

```python
sub = df.dropna(subset=["urban_population_pct", "internet_users_pct"])
sub.plot.scatter(x="urban_population_pct", y="internet_users_pct", s=6, alpha=0.3, figsize=(8, 5))
```
