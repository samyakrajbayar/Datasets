# USGS Water-Monitoring Site Inventory

Starter notebook for the Kaggle dataset in this folder. It loads `usgs_sites.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `usgs_sites.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("usgs_sites.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Sites by type

```python
df["site_type"].value_counts().head(15).to_frame("sites")
```

## 5. Sites per state

```python
df["state_name"].value_counts().head(20).to_frame("sites")
```

## 6. Altitude distribution

```python
df["altitude_ft"].dropna().clip(upper=10000).hist(bins=50, figsize=(8, 4))
```

## 7. Site-type flags

```python
for c in ["is_stream", "is_groundwater", "is_lake", "is_spring", "is_estuary"]:
    print(c, int((df[c].astype(str).str.lower() == "true").sum()))
```
