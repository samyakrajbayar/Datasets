# Hacker News Stories 2023–2026

Starter notebook for the Kaggle dataset in this folder. It loads `hacker_news_stories.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `hacker_news_stories.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("hacker_news_stories.csv", low_memory=False, parse_dates=["created_at"])
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Top domains

```python
df["domain"].value_counts().head(20).to_frame("stories")
```

## 5. Post types

```python
pd.Series({"ask_hn": (df["is_ask_hn"] == "true").sum(),
            "show_hn": (df["is_show_hn"] == "true").sum(),
            "external_link": (df["has_url"] == "true").sum()})
```

## 6. Points distribution

```python
df["points"].dropna().clip(upper=1000).hist(bins=50, figsize=(8, 4))
```

## 7. Monthly volume

```python
df.groupby(df["created_at"].dt.to_period("M")).size().plot(figsize=(11, 4))
```

## 8. Hour of day

```python
df.groupby("hour")["object_id"].count().plot(kind="bar", figsize=(10, 4))
```
