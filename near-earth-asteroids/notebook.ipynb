# Near-Earth Asteroids

Starter notebook for the Kaggle dataset in this folder. It loads `near_earth_asteroids.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `near_earth_asteroids.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("near_earth_asteroids.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Orbit classes

```python
df["orbit_class"].value_counts().to_frame("objects")
```

## 5. Semi-major axis vs eccentricity

```python
df.plot.scatter(x="semi_major_axis_au", y="eccentricity", s=3, alpha=0.25, figsize=(8, 5))
```

## 6. Size distribution

```python
sizes = df["diameter_km"].dropna()
import numpy as np
np.logspace(0, 3, 40)  # reference decades
sizes[sizes > 0].plot(kind="hist", bins=np.logspace(-3, 1.4, 40), logy=True, figsize=(10, 4),
                      title="Diameter distribution (km, log bins)")
```

## 7. Potentially hazardous objects

```python
df.groupby("orbit_class")["is_potentially_hazardous"].apply(
    lambda s: (s.astype(str).str.lower() == "true").mean() * 100).round(1).to_frame("pha_%")
```

## 8. Closest approaches

```python
df.nsmallest(15, "earth_moid_au")[["full_name", "earth_moid_au", "h_magnitude_size_m", "orbit_class"]]
```

## 9. Data completeness by size

```python
df["size_bin"] = pd.cut(df["diameter_km"], [0, .1, .5, 1, 5, 50])
df.groupby("size_bin", observed=True)[["has_albedo", "has_spectral_type"]].apply(
    lambda t: t.apply(lambda s: (s.astype(str).str.lower() == "true").mean()) * 100).round(1)
```
