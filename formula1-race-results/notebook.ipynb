# Formula 1 Race Results 1950–2026

Starter notebook for the Kaggle dataset in this folder. It loads `f1_results.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `f1_results.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("f1_results.csv", low_memory=False, parse_dates=["race_date"])
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Races per season

```python
df.groupby("season")["round"].nunique().tail(30).plot(kind="bar", figsize=(11, 4))
```

## 5. Top drivers by wins

```python
df[df["is_win"] == "true"]["driver_id"].value_counts().head(15).to_frame("wins")
```

## 6. Constructors

```python
df["constructor_name"].value_counts().head(12).to_frame("results")
```

## 7. Grid vs finish

```python
sub = df[(df["grid"] > 0) & df["position"].notna()]
sub.plot.scatter(x="grid", y="position", s=4, alpha=0.15, figsize=(7, 7))
```

## 8. Status mix

```python
df["status"].value_counts().head(15).to_frame("results")
```
