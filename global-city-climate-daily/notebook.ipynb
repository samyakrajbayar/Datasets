# Global City Climate 2015-2025: 192K Daily Records — Exploratory Data Analysis

Welcome to the starter notebook for the **Global City Climate 2015-2025: 192K Daily Records** dataset!
This notebook loads the dataset, inspects summary statistics, and provides baseline visualizations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Set plotting aesthetics
sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams['figure.figsize'] = (10, 6)
```

## 1. Load Dataset
We read the primary dataset file and preview the first few rows.

```python
file_path = 'city_climate_daily.csv'
if not os.path.exists(file_path):
    file_path = os.path.join('/kaggle/input/global-city-climate-daily', 'city_climate_daily.csv')

df = pd.read_csv(file_path, low_memory=False)
print(f'Dataset Dimensions: {df.shape[0]:,} rows, {df.shape[1]} columns')
df.head()
```

## 2. Dataset Summary & Missing Values
Overview of data types and null counts across features.

```python
df.info()
null_counts = df.isnull().sum()
null_pct = (null_counts / len(df)) * 100
missing_df = pd.DataFrame({'Missing Count': null_counts, 'Percent (%)': null_pct})
missing_df[missing_df['Missing Count'] > 0].sort_values(by='Percent (%)', ascending=False).head(15)
```

## 3. Numerical Summary Statistics
Summary statistics for numerical features.

```python
df.describe()
```

## 4. Initial Distribution Visualizations
Plot distributions for available numerical variables.

```python
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
if len(numeric_cols) > 0:
    cols_to_plot = numeric_cols[:min(4, len(numeric_cols))]
    df[cols_to_plot].hist(bins=25, figsize=(12, 8), layout=(2, 2))
    plt.suptitle('Distributions of Numeric Features', fontsize=14)
    plt.tight_layout()
    plt.show()
else:
    print('No numeric columns found for histogram plots.')
```
