# Sequenced Genomes — NCBI RefSeq Assembly Catalogue

Starter notebook for the Kaggle dataset in this folder. It loads `genome_assemblies.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `genome_assemblies.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("genome_assemblies.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Assemblies by organism group

```python
df["group"].value_counts().head(12).plot(kind="bar", figsize=(11, 4))
```

## 5. Genome size distribution

```python
df["genome_size_mb"].dropna().clip(upper=5000).hist(bins=80, figsize=(10, 4))
```

## 6. Assembly level over time

```python
pd.crosstab(pd.to_numeric(df["seq_rel_year"], errors="coerce"), df["assembly_level"]).tail(20).plot(figsize=(11, 4))
```

## 7. Largest genomes

```python
df.nlargest(12, "genome_size_mb")[["organism_name", "assembly_level", "genome_size_mb", "total_gene_count"]]
```

## 8. GC content by group

```python
df.boxplot(column="gc_percent", by="group", figsize=(13, 4), rot=90)
```

## 9. Gene density

```python
df[["genes_per_mb", "genome_size_mb"]].dropna().sample(5000, random_state=0).plot.scatter(x="genome_size_mb", y="genes_per_mb", alpha=0.2, logx=True)
```
