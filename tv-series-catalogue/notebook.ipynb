# Television Series Catalogue — TVmaze

Starter notebook for the Kaggle dataset in this folder. It loads `tv_series.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `tv_series.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("tv_series.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Shows by type

```python
df["type"].value_counts().head(12).plot(kind="bar", figsize=(10, 4))
```

## 5. Rating distribution

```python
df["rating_average"].dropna().hist(bins=40, figsize=(9, 4))
```

## 6. Top rated shows with at least 500 votes of weight

```python
df[(df["weight"] >= 90) & df["rating_average"].notna()].nlargest(15, "rating_average")[["name", "network_name", "premiered_year", "rating_average", "weight"]]
```

## 7. Premières per year

```python
pd.to_numeric(df["premiered_year"], errors="coerce").dropna().astype(int).value_counts().sort_index().loc[1950:].plot(figsize=(11, 4))
```

## 8. Genre mix over time

```python
pd.crosstab(pd.to_numeric(df["premiered_year"], errors="coerce").floordiv(10).mul(10), df["genre_drama"].astype(str).str.lower().eq("true")).tail(8)
```

## 9. Networks

```python
df["network_name"].value_counts().head(15)
```

## 10. Runtime by genre

```python
df.boxplot(column="runtime_minutes", by="type", figsize=(11, 4), rot=45)
```
