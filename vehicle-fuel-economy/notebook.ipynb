# U.S. Vehicle Fuel Economy & Emissions

Starter notebook for the Kaggle dataset in this folder. It loads `fuel_economy.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `fuel_economy.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("fuel_economy.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Combined MPG distribution

```python
df["comb08"].dropna().clip(upper=80).hist(bins=60, figsize=(8, 4))
```

## 5. Median MPG by model year

```python
df.groupby("year")["comb08"].median().tail(30).plot(figsize=(10, 4))
```

## 6. Powertrain mix

```python
for c in ["is_electric", "is_hybrid", "is_plugin_hybrid", "is_diesel", "is_flex_fuel"]:
    print(c, round((df[c].astype(str).str.lower() == "true").mean() * 100, 2), "%")
```

## 7. CO2 vs MPG

```python
df.dropna(subset=["comb08", "co2"]).sample(5000, random_state=0).plot.scatter("comb08", "co2", alpha=0.3, figsize=(8, 5))
```
