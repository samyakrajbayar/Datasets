# FIDE Chess Ratings

Starter notebook for the Kaggle dataset in this folder. It loads `fide_players.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `fide_players.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("fide_players.csv", dtype={"fide_id": str}, low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Rating distribution

```python
df.loc[df["rating"] > 0, "rating"].hist(bins=60, figsize=(9, 4))
```

## 5. Top federations

```python
df["country"].value_counts().head(15).to_frame("players")
```

## 6. Titles

```python
df["title"].value_counts(dropna=False).head(10).to_frame("players")
```

## 7. Rating by age

```python
sub = df.dropna(subset=["age", "rating"]).query("rating > 0 and age < 90")
sub.groupby("age")["rating"].mean().plot(figsize=(10, 4))
```

## 8. Elite players

```python
df.nlargest(15, "rating")[["name", "country", "rating", "birth_year", "title"]]
```
