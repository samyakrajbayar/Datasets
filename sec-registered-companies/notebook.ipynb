# SEC-Registered Companies

Starter notebook for the Kaggle dataset in this folder. It loads `sec_companies.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `sec_companies.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("sec_companies.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Entity types

```python
df["entity_type"].value_counts().head(10).to_frame("companies")
```

## 5. Industry divisions

```python
df["sic_division"].value_counts().to_frame("companies")
```

## 6. Filing activity

```python
df["days_since_latest_filing"].dropna().clip(upper=3000).plot(
    kind="hist", bins=50, figsize=(10, 4), title="Days since latest filing")
```

## 7. Filing mix by division

```python
df.groupby("sic_division")[["filings_10k", "filings_8k", "filings_form4"]].mean().round(1)
```

## 8. Foreign issuers

```python
df.groupby("is_foreign_issuer").size().to_frame("companies")
```

## 9. Largest filers by recent filings

```python
df.nlargest(15, "filing_count_recent")[["name", "sic_description", "filing_count_recent", "latest_filing_date"]]
```

## 10. Annual-report lag

```python
df["annual_report_lag_days"].dropna().describe().round(1).to_frame("days")
```
