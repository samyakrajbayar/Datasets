# US Public Schools — NCES Common Core of Data Directory

Starter notebook for the Kaggle dataset in this folder. It loads `us_public_schools.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `us_public_schools.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("us_public_schools.csv", low_memory=False, dtype={"ncessch": str, "leaid": str, "zip_location": str})
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Schools per state

```python
df["state_abbr"].value_counts().head(20).plot(kind="bar", figsize=(11, 4))
```

## 5. Enrolment distribution

```python
df["enrollment"].dropna().clip(upper=4000).hist(bins=60, figsize=(10, 4))
```

## 6. Student-teacher ratio by locale

```python
df.groupby("locale_type")["student_teacher_ratio"].median().sort_values()
```

## 7. Free lunch share vs enrolment

```python
df[["enrollment", "free_or_reduced_lunch_pct"]].dropna().sample(8000, random_state=0).plot.scatter(x="enrollment", y="free_or_reduced_lunch_pct", alpha=0.15, logx=True)
```

## 8. School types

```python
df["school_type_desc"].value_counts()
```

## 9. Charter share by state

```python
ch = df.assign(charter=df["is_charter"].astype(str).str.lower().eq("true"))
ch.groupby("state_abbr")["charter"].mean().sort_values(ascending=False).head(15)
```

## 10. Map of schools (sample)

```python
df.sample(3000, random_state=1).plot.scatter(x="longitude", y="latitude", s=2, alpha=0.3, figsize=(11, 6))
```
