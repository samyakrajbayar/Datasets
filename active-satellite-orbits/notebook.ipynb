# Active Satellite Orbits

Starter notebook for the Kaggle dataset in this folder. It loads `satellites.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `satellites.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("satellites.csv", low_memory=False)
for c in ["mean_altitude_km", "inclination", "period", "mean_motion"]:
    df[c] = pd.to_numeric(df[c], errors="coerce")
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Orbit shells

Inclination versus altitude — the mega-constellation shells are obvious.

```python
df.dropna(subset=["inclination", "mean_altitude_km"]).query("mean_altitude_km < 40000").plot.scatter(
    x="inclination", y="mean_altitude_km", s=3, alpha=0.3, figsize=(11, 6))
```

## 5. Largest constellations

```python
df["constellation"].value_counts().head(12).plot.barh(figsize=(8, 5))
```

## 6. Orbital regimes

```python
df["orbital_regime"].value_counts()
```

## 7. Altitude histogram for Starlink

```python
df.query("constellation == 'Starlink'")["mean_altitude_km"].plot.hist(bins=80, figsize=(10, 4))
```

## 8. Launch waves

```python
df["launch_year"].dropna().astype(int).value_counts().sort_index().plot.bar(figsize=(12, 4))
```

## 9. Sun-synchronous fleet by operator

```python
df[df["is_sun_synchronous"].astype(str).str.lower() == "true"]["operator_group"].value_counts().head(10)
```
