# U.S. Federal Contract Awards — FY2025–FY2026

Starter notebook for the Kaggle dataset in this folder. It loads `federal_awards.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `federal_awards.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("federal_awards.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Top agencies by obligations

```python
df.groupby("awarding_agency")["award_amount"].sum().sort_values(ascending=False).head(15).div(1e9).round(2).to_frame("bn_usd")
```

## 5. Top recipients

```python
df.groupby("recipient_name")["award_amount"].sum().sort_values(ascending=False).head(15).div(1e9).round(2).to_frame("bn_usd")
```

## 6. Award size distribution

```python
df["award_amount"].dropna().clip(upper=5e8).hist(bins=50, figsize=(8, 4))
```

## 7. NAICS sectors

```python
df["naics_sector"].value_counts().head(15).to_frame("awards")
```
