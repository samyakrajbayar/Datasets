# Global Weather Station Network

Starter notebook for the Kaggle dataset in this folder. It loads `weather_stations.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `weather_stations.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("weather_stations.csv", dtype={"station_id": str}, low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Stations by country

```python
df["country_name"].value_counts().head(15).to_frame("stations")
```

## 5. Record length distribution

```python
df["record_years"].dropna().clip(upper=150).hist(bins=50, figsize=(8, 4))
```

## 6. Element coverage

```python
for c in ["has_precipitation", "has_temperature", "has_snow", "has_wind"]:
    print(c, round((df[c].astype(str).str.lower() == "true").mean() * 100, 1), "%")
```

## 7. Map

```python
sub = df.dropna(subset=["latitude", "longitude"])
sub.plot.scatter(x="longitude", y="latitude", s=1, alpha=0.3, figsize=(11, 6))
```

## 8. Active stations by decade started

```python
(df["first_year"] // 10 * 10).value_counts().sort_index().tail(15).plot(kind="bar", figsize=(10, 4))
```
