# Steam Games Market

Starter notebook for the Kaggle dataset in this folder. It loads `steam_games.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `steam_games.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("steam_games.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Price distribution

```python
df["price_usd"].dropna().clip(upper=80).plot(
    kind="hist", bins=40, figsize=(10, 4), title="Price distribution (clipped at $80)")
```

## 5. Free vs paid

```python
df.groupby("is_free")[["owners_mid", "total_reviews", "ccu"]].median().round(1)
```

## 6. Biggest games by estimated owners

```python
df.nlargest(15, "owners_mid")[["name", "developer", "owners_mid", "review_ratio_pct", "price_usd"]]
```

## 7. Review ratio vs price band

```python
df.groupby("price_band")["review_ratio_pct"].agg(["count", "median"]).round(1)
```

## 8. Engagement: median vs average playtime

```python
sub = df.dropna(subset=["median_playtime_hours", "average_playtime_hours"]).query("average_playtime_hours < 200")
sub.plot.scatter(x="average_playtime_hours", y="median_playtime_hours", s=5, alpha=0.3, figsize=(8, 5))
```

## 9. Top publishers by title count

```python
df["publisher"].value_counts().head(15).to_frame("titles")
```
