# MLB Player Seasons 2015–2025

Starter notebook for the Kaggle dataset in this folder. It loads `mlb_player_seasons.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `mlb_player_seasons.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("mlb_player_seasons.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Rows by group

```python
df["group"].value_counts().to_frame("rows")
```

## 5. Top batting seasons

```python
df[df["group"] == "hitting"].nlargest(10, "home_runs")[["player_name", "season", "home_runs", "rbi", "ops"]]
```

## 6. ERA distribution

```python
df.loc[df["group"] == "pitching", "era"].apply(pd.to_numeric, errors="coerce").dropna().clip(upper=10).hist(bins=40, figsize=(8, 4))
```

## 7. OPS vs home runs

```python
sub = df[df["group"] == "hitting"].dropna(subset=["ops", "home_runs"])
sub.plot.scatter(x="home_runs", y="ops", s=5, alpha=0.25, figsize=(8, 5))
```

## 8. Age curve

```python
sub = df[(df["group"] == "hitting") & df["age"].notna()]
sub.groupby("age")["ops"].mean().loc[20:40].plot(figsize=(9, 4))
```
