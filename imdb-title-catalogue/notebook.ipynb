# IMDb Title Catalogue 1990–2026

Starter notebook for the Kaggle dataset in this folder. It loads `imdb_titles.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `imdb_titles.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("imdb_titles.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Titles by type

```python
df["title_type"].value_counts().to_frame("titles")
```

## 5. Rating distribution

```python
df["average_rating"].dropna().hist(bins=40, figsize=(8, 4))
```

## 6. Ratings by decade

```python
df.groupby("decade")["average_rating"].agg(["mean", "count"]).round(2)
```

## 7. Genre ratings

```python
for g in ["is_drama", "is_comedy", "is_documentary", "is_horror", "is_action"]:
    sub = df[df[g].astype(str).str.lower() == "true"]
    print(g, round(sub["average_rating"].mean(), 2), len(sub))
```

## 8. Runtime vs rating

```python
sub = df.dropna(subset=["runtime_minutes", "average_rating"])
sub.plot.scatter(x="runtime_minutes", y="average_rating", s=4, alpha=0.15, figsize=(8, 5))
```
