# Human Gene Catalogue — HGNC Complete Set

Starter notebook for the Kaggle dataset in this folder. It loads `hgnc_genes.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `hgnc_genes.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("hgnc_genes.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Locus groups

```python
df["locus_group"].value_counts().to_frame("genes")
```

## 5. Genes per chromosome

```python
df["chromosome"].value_counts().reindex([str(i) for i in range(1, 23)] + ["X", "Y", "MT"]).to_frame("genes")
```

## 6. Cross-reference coverage

```python
cols = ["has_ensembl", "has_uniprot", "has_omim", "has_refseq", "has_ccds", "has_mane_select"]
for c in cols:
    print(c, round((df[c].astype(str).str.lower() == "true").mean() * 100, 1), "%")
```

## 7. Largest gene groups

```python
df.loc[df["gene_group"].notna(), "gene_group"].str.split("|").explode().value_counts().head(15)
```
