# arXiv Preprints

Starter notebook for the Kaggle dataset in this folder. It loads `arxiv_papers.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `arxiv_papers.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("arxiv_papers.csv", low_memory=False)
df["published"] = pd.to_datetime(df["published"], errors="coerce")
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Papers by primary category

```python
df["primary_category"].value_counts().head(15).to_frame("papers")
```

## 5. Authors per paper

```python
df["author_count"].describe().round(2).to_frame("authors")
```

## 6. Abstract length by archive

```python
df.groupby("primary_group")["abstract_words"].median().sort_values().to_frame("median_words")
```

## 7. Submissions per day

```python
df.set_index("published").resample("D").size().plot(figsize=(11, 4), title="Papers per day in the sample")
```

## 8. Published in a journal?

```python
df.groupby("primary_group")[["has_journal_ref", "has_doi"]].apply(
    lambda t: t.apply(lambda s: (s.astype(str).str.lower() == "true").mean()).mul(100).round(1))
```

## 9. Most common title words

```python
from collections import Counter
words = Counter(w.lower() for t in df["title"].dropna() for w in t.split() if len(w) > 3)
pd.Series(dict(words.most_common(20))).sort_values().plot.barh(figsize=(8, 6))
```
