# FDA Drug Labels — SPL Section Corpus

Starter notebook for the Kaggle dataset in this folder. It loads `drug_labels.csv` with pandas and walks through quality checks and a first piece of analysis.

**Files:** `drug_labels.csv`, `data_dictionary.csv`.

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv("drug_labels.csv", low_memory=False)
print(df.shape)
```

## 2. Shape, types and missing values

```python
print(f"{df.shape[0]:,} rows x {df.shape[1]} columns")

missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0].round(1).to_frame("missing_%").head(15)
```

## 3. Numeric summary

```python
df.describe(include='number').T.round(2).head(20)
```

## 4. Product types

```python
df["product_type"].value_counts().to_frame("labels")
```

## 5. Section coverage

```python
df["section_coverage"].value_counts().sort_index().to_frame("labels")
```

## 6. Most common routes

```python
df["route"].str.split(";").explode().str.strip().value_counts().head(15)
```

## 7. Label age

```python
df["label_age_years"].dropna().clip(upper=30).hist(bins=30, figsize=(8, 4))
```
